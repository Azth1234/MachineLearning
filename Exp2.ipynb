```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.bar(df[" Name "],df["TotalMarks"])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df[" Attendance "], bins=5)
plt.xlabel("Attendance distributuiin")
plt.ylabel("Frequency")
plt.show()
```

```python
gender_count=df[" Gender "].value_counts()

plt.figure(figsize=(8,5))
plt.pie(gender_count, labels=gender_count.index)
plt.title("gender distributuiin")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df[" Attendance "],df["TotalMarks"])
plt.title("Attendance vs TotalMarks")
plt.xlabel("Attendance")
plt.ylabel("TotalMarks")
plt.show()
```
